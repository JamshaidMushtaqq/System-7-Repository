# 1. Conditions — if, elif, else

```python
marks = 75

if marks >= 80:
    print("Grade: A")
elif marks >= 60:
    print("Grade: B")
else:
    print("keep improving")
```

# 2. Loops — for aur while

```python
# for loop
for i in range(1,6):
    print(i,end=" ")
```

```python
# while loop
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1
    
```

```python
# range function
print(list(range(1, 6)))   
print(list(range(2, 11, 2)))  
```

# Dictionary

```python
student = {
    "name": "Ali",
    "marks": 85,
    "city": "Lahore"
}

print(student["name"])  

student["marks"] = 90
student["grade"] = "A"

print(student)
```

# File handling

```python
with open("notes.txt", "w") as file:
    file.write("I am learning Python!")
```

```python
with open("notes.txt", "r") as file:
    content = file.read()
    print(content)
```

# Exceptions — errors

```python
try:
    age = int(input("Enter your age: "))
    print("Your age is:", age)
except ValueError:
    print("Please enter a valid whole number.")
```
