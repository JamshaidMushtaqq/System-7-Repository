```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

df = pd.DataFrame({
    "area":      [1200, 1800, 900, 2500, 1500, 2000],
    "bedrooms":  [2, 3, 1, 4, 3, 3],
    "city":      ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Islamabad"],
    "furnished": ["yes", "no", "no", "yes", "yes", "no"],
    "price":     [60, 85, 40, 150, 80, 110],
})
X = df.drop(columns="price")
y = df["price"]

num_cols = ["area", "bedrooms"]
cat_cols = ["city", "furnished"]

pre = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

# Ek step mein dono kaam
X_new = pre.fit_transform(X)
print("Before:", X.shape, "| After:", X_new.shape)
print(pre.get_feature_names_out())

# Model ke sath pipeline mein
model = Pipeline([("pre", pre), ("lr", LinearRegression())])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
model.fit(X_train, y_train)

# Naya ghar, jis ka city (Multan) train mein nahi tha
new_house = pd.DataFrame({"area": [1600], "bedrooms": [3],
                          "city": ["Multan"], "furnished": ["yes"]})
print("Prediction:", model.predict(new_house).round(1))
```
