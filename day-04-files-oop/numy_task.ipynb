# Numpy

```python
# Perform NumPy array operations on a sample matrix
import numpy as np
```

**General Questions**

```python

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
shape = data.shape
shape
```

```python
#Find the average age of students.
average_age = np.mean(data[:, 0])
average_age
```

```python
#Extract Math marks of all students.
math_marks = data[:, 1]
math_marks
```

```python
#Find the highest Science mark.
highest_science_mark = np.max(data[:, 2])
highest_science_mark
```

```python
#Get details of the student who scored more than 90 in Math.
student_details = data[data[:, 1] > 90, 1]
student_details
```

```python
#Increase Math marks of all students by 5.
data[:, 1] += 5
data[:, 1]
```

```python
#Find how many students are younger than 19.
num_students_younger_than_19 = np.sum(data[:, 0] < 19)
num_students_younger_than_19
```

```python
#Calculate the average marks in each subject (column-wise mean).
average_marks = np.mean(data[:, 1:3], axis=0)
average_marks
```

```python
#Get data of students who scored at least 80 in both subjects.
student_data = data[(data[:, 1] >= 80) & (data[:, 2] >= 80)]
student_data
```

```python
#Replace all Science marks < 75 with 0.
data[:, 2] = np.where(data[:, 2] < 75, 0, data[:, 2])
data[:, 2]
```
