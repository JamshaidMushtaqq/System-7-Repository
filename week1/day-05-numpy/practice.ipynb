# Pandas Practice

```python
# series
import pandas as pd 

marks = pd.Series([90, 80, 70, 60, 50], index=['Math', 'Science', 'English', 'History', 'Art'])
print(marks)
```

```python
# dataframe
data =  {
    "Name": ["jamshaid", "ali", "ahmed", "hassan", "bilal"],
    "Marks": [90, 80, 70, 60, 50],
    "age": [20, 21, 22, 23, 24]
}

df = pd.DataFrame(data)
print(df)
```

```python
#filtering data
# students with marks greater or equal 80
print(df[df['Marks'] >= 80])
```

```python
# sorting data
print(df.sort_values(by="Marks", ascending=False))
```

```python
# handle missing values
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Marks": [85, None, 78, 92],
    "Age": [20, 21, None, 22]
}

df = pd.DataFrame(data)

# Missing values check karo
print(df.isnull())

# Har column mein missing values count karo
print(df.isnull().sum())

# Missing marks ko 0 se replace karo
df["Marks"] = df["Marks"].fillna(0)

print(df)
```
