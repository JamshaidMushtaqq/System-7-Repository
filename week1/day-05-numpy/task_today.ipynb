```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Marks": [85, None, 78, 92, None],
    "Attendance": [90, 85, None, 95, 80]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
df[["Marks", "Attendance"]].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Before Cleaning")
plt.xlabel("Student Index")
plt.ylabel("Values")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df_clean = df.copy()

df_clean["Marks"] = df_clean["Marks"].fillna(
    df_clean["Marks"].mean()
)

df_clean["Attendance"] = df_clean["Attendance"].fillna(
    df_clean["Attendance"].mean()
)

print("Cleaned Dataset:")
print(df_clean)

print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())
```

```python
df_clean[["Marks", "Attendance"]].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("After Cleaning")
plt.xlabel("Student Index")
plt.ylabel("Values")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(1,2, figsize=(12, 5))

df[["Marks", "Attendance"]].plot(
    kind = "bar", ax=axes[0])

axes[0].set_title("before cleaning")
axes[0].set_xlabel("Student Index")
axes[0].set_ylabel("Values")

df_clean[["Marks", "Attendance"]].plot(
    kind = "bar", ax=axes[1])

axes[1].set_title("after cleaning")
axes[1].set_xlabel("Student Index")
axes[1].set_ylabel("Values")

plt.tight_layout()
plt.show()
```
