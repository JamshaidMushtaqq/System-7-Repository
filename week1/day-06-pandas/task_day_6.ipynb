```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
# load dataset
df = pd.read_csv("StudentsPerformance.csv")
```

# Analyze the csv file 

```python
print("first 5 rows")
print(df.head())

print("\nshape of dataset")
print(df.shape)

print("column names")
df.columns.tolist()
```

```python
print("information about dataset")
print(df.info())
```

```python
print("check missing values")
print(df.isnull().sum())
```

```python
print("check duplicates")
print(df.duplicated().sum())
```

# statistics summary

```python
scores = ["math score","reading score","writing score"]
print("summary statistics")
print(df[scores].describe().round(2))
```

```python
print("mean scores")
print(df[scores].mean().round(2))
```

```python
print("meadian score")
print(df[scores].median().round(2))
```

```python
# chart1 avarage marks by subject
avarage_scores = df[scores].mean()
avarage_scores.plot(kind="bar", figsize=(8,5))
plt.title("Average marks by subject")
plt.xlabel("Subjects")
plt.ylabel("Avarage Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# charts 2 Marks distribution
df["math score"].plot(kind="hist",bins=10,figsize=(8,5),edgecolor= 'black')
plt.title("Distribution of math marks")
plt.xlabel("Math Marks")
plt.ylabel("Number of students")
plt.tight_layout()
plt.show()
```

```python
# chart 3 math and reading realation

plt.figure(figsize=(8,5))
plt.scatter(df["math score"],df["reading score"])
plt.title("Math Score VS Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.tight_layout()
plt.show()
```

```python
# comparison of test preparation
prep_scores = df.groupby("test preparation course")[scores].mean().round(2)

prep_scores.plot(kind='bar',figsize=(9,5))
plt.title("scores by test prepration")
plt.xlabel("Test prepration")
plt.ylabel("Average Marks")
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```
